# panchi 2.0 — feature tour

A quick tour of what's new in **2.0**: rich `_repr_latex_` display, notebook-native animation playback, and the two new visualization features (transforming your own vectors, and comparing multiple spans).

> This notebook ships **pre-run** so you can read it without a kernel. Re-run it in Jupyter to reproduce everything live.

## 1. Rich display — objects render as LaTeX

Every primitive and result object defines `_repr_latex_`, so Jupyter shows real math. `str()`/`repr()` are unchanged.

```python
from fractions import Fraction
from panchi import Vector, Matrix, VectorSpace

Matrix([[2, Fraction(1, 2), 0], [1, 3, -1], [0, 4, 1]])
```

The centerpiece — a full **row reduction** renders as an arrow chain:

```python
from panchi.algorithms import rref
rref(Matrix([[1, 2, 1], [2, 4, 0], [1, 1, 1]]))
```

```python
from panchi.algorithms import eigen
eigen(Matrix([[2, 0], [0, 3]]))
```

```python
from panchi.algorithms import inverse
inverse(Matrix([[4, 7], [2, 6]]))
```

_See `latex_showcase.html` for the full gallery of rich-display types._

## 2. Notebook-native animation playback

Inside a notebook, `animate_*` returns an object that plays **inline** (matplotlib's own JS player via `to_jshtml`) — no more static final frame, no IPython dependency. Outside a notebook it behaves exactly as before.

Below: transforming three custom vectors (feature #1).

```python
from panchi.visualizations import Animator2D

anim = Animator2D()
anim.animate_transform(
    Matrix([[2, 1], [0, 1]]),
    vectors=[Vector([1, 0]), Vector([1, 2]), Vector([-1, 1])],
)
```
